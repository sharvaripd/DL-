```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Machine learning libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
from sklearn.metrics import accuracy_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

```

```python
import pandas as pd
df=pd.read_csv("FridayWorkingHoursAfternoon.csv")
df.head()
```

```python
x= df.iloc[:,0:71]
y= df.iloc[:,71]
print(x.head())
print(y.head())
```

```python
#Train Test Split of the data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.shape,x_test.shape)
print(y_train.shape,y_test.shape)
```

```python
#Feature Scaling
sc=StandardScaler()

# Handle NaN and Inf values in x_train and x_test before scaling
# Convert all infinite values to NaN first to unify missing value handling
x_train.replace([np.inf, -np.inf], np.nan, inplace=True)
x_test.replace([np.inf, -np.inf], np.nan, inplace=True)

# Fill remaining NaN values with the mean of their respective columns.
x_train = x_train.fillna(x_train.mean(numeric_only=True))
x_test = x_test.fillna(x_test.mean(numeric_only=True))

x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
# Building ANN
ann=Sequential()
ann.add(Dense(units=64,activation='tanh')) # Unit represents the number of perceptrons
ann.add(Dense(units=32,activation='tanh')) # Hidden Layer
ann.add(Dense(units=1,activation='sigmoid')) # Output Layer
```

```python
#Compile
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode the categorical labels in y_train and y_test to numerical format
# The labels are 'BENIGN' and 'DDoS', which need to be converted to 0 and 1 for binary_crossentropy.
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test) # Use transform on test set to apply the same mapping

history=ann.fit(x_train,y_train_encoded,batch_size=32,epochs=10,validation_split=0.20)
```
